## 2. Mathematical Foundations

This section introduces the mathematical concepts required to understand the experiments in the Probability Paradox Lab. These concepts include probability, conditional probability, independent events, Bayes' theorem, and Monte Carlo simulation.

### 2.1 Sample Spaces and Events

A **sample space** is the set of all possible outcomes of a random experiment. It is commonly denoted by \(\Omega\).

For example, when rolling a fair six-sided die, the sample space is

$$
\Omega = \{1,2,3,4,5,6\}
$$

An **event** is a subset of the sample space. For example, the event of rolling an even number is

$$
A = \{2,4,6\}
$$

If all outcomes are equally likely, the probability of an event \(A\) is calculated as

$$
P(A) = \frac{|A|}{|\Omega|}
$$

where \(|A|\) represents the number of outcomes belonging to event \(A\), and \(|\Omega|\) represents the total number of possible outcomes.

For the event of rolling an even number,

$$
P(A) = \frac{3}{6} = \frac{1}{2}
$$

Therefore, the probability of rolling an even number is \(0.5\), or \(50\%\).

### 2.2 Fundamental Rules of Probability

Probability values range from \(0\) to \(1\):

$$
0 \leq P(A) \leq 1
$$

A probability of \(0\) represents an impossible event, while a probability of \(1\) represents a certain event.

The probability of the entire sample space is

$$
P(\Omega) = 1
$$

The probability of the complement of an event \(A\), denoted by \(A^c\), is

$$
P(A^c) = 1 - P(A)
$$

For two events \(A\) and \(B\), the addition rule is

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

Here, \(A \cup B\) means that at least one of the events occurs, while \(A \cap B\) means that both events occur.

If \(A\) and \(B\) are mutually exclusive, they cannot occur simultaneously, so

$$
P(A \cap B) = 0
$$

Consequently,

$$
P(A \cup B) = P(A) + P(B)
$$

### 2.3 Conditional Probability

**Conditional probability** measures the probability of an event occurring when another event is already known to have occurred.

The conditional probability of \(A\) given \(B\) is defined as

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

provided that \(P(B) > 0\).

This formula is particularly important when new information changes the probabilities assigned to possible outcomes.

For example, in the Monty Hall problem, the host's decision to reveal a goat provides additional information. The probability of winning after switching doors is therefore different from the probability associated with the initial choice.

### 2.4 Independent Events

Two events \(A\) and \(B\) are independent if the occurrence of one does not change the probability of the other.

Mathematically, independence means

$$
P(A \cap B) = P(A)P(B)
$$

Equivalently, when \(P(B) > 0\),

$$
P(A \mid B) = P(A)
$$

For example, two separate tosses of a fair coin are independent, assuming the coin tosses do not influence one another.

Independence is an important assumption in many probability models, but it must be justified rather than assumed automatically.

### 2.5 Bayes' Theorem

Bayes' theorem provides a way to update the probability of a hypothesis after observing new evidence.

For events \(A\) and \(B\), where \(P(B) > 0\), Bayes' theorem is

$$
P(A \mid B) =
\frac{P(B \mid A)P(A)}{P(B)}
$$

The terms have the following meanings:

- \(P(A)\): the prior probability of event \(A\).
- \(P(B \mid A)\): the probability of observing evidence \(B\) when \(A\) is true.
- \(P(B)\): the overall probability of observing evidence \(B\).
- \(P(A \mid B)\): the posterior probability of \(A\) after observing \(B\).

If \(A\) and \(A^c\) represent complementary hypotheses, the denominator can be expanded as

$$
P(B) = P(B \mid A)P(A)
+ P(B \mid A^c)P(A^c)
$$

Therefore,

$$
P(A \mid B) =
\frac{P(B \mid A)P(A)}
{P(B \mid A)P(A) + P(B \mid A^c)P(A^c)}
$$

This formula will be used in the third case study to investigate how disease prevalence, test sensitivity, and false-positive rates influence the probability that a person actually has a condition after receiving a positive test result.

### 2.6 Monte Carlo Simulation

A **Monte Carlo simulation** estimates probabilities by repeatedly performing a random experiment and measuring how frequently a particular event occurs.

Suppose an experiment is repeated \(N\) times, and event \(A\) occurs \(K\) times. The estimated probability of \(A\) is

$$
\hat{P}(A) = \frac{K}{N}
$$

where:

- \(N\) is the total number of trials.
- \(K\) is the number of trials in which event \(A\) occurs.
- \(\hat{P}(A)\) is the estimated probability.

For example, if an event occurs \(4{,}970\) times in \(10{,}000\) trials, its estimated probability is

$$
\hat{P}(A) = \frac{4970}{10000} = 0.497
$$

If the theoretical probability is \(P(A) = 0.5\), the estimate is close to the theoretical value.

However, Monte Carlo simulation does not generally guarantee that an estimate will equal the theoretical probability exactly. Random variation causes estimates to fluctuate around the theoretical value, especially when the number of trials is small.

As the number of trials increases, the estimated probability generally becomes more stable and tends to approach the theoretical probability.

```python
import numpy as np

# Use a fixed seed so the experiment is reproduciblex§
rng = np.random.default_rng()

# Number of simulated coin tosses
n_trials = 10000

# Simulate a fair coin: 1 represents heads, 0 represents tails
tosses = rng.choice([0, 1], size=n_trials)

# Count the number of heads
heads_count = np.count_nonzero(tosses == 1)

# Estimate the probability of heads
estimated_probability = heads_count / n_trials

# Theoretical probability for a fair coin
theoretical_probability = 0.5

print(f"Number of trials: {n_trials:,}")
print(f"Number of heads: {heads_count:,}")
print(f"Estimated probability: {estimated_probability:.4f}")
print(f"Theoretical probability: {theoretical_probability:.4f}")
print(f"Absolute error: {abs(estimated_probability - theoretical_probability):.4f}")
```

```python
import matplotlib.pyplot as plt

# Calculate the cumulative number of heads
cumulative_heads = np.cumsum(tosses)

# Number of tosses performed at each point
trial_numbers = np.arange(1, n_trials + 1)

# Estimate the probability of heads after each toss
estimated_probabilities = cumulative_heads / trial_numbers

# Plot the estimated probability
plt.figure(figsize=(10, 5))
plt.plot(
    trial_numbers,
    estimated_probabilities,
    label="Estimated probability"
)

# Show the theoretical probability
plt.axhline(
    y=theoretical_probability,
    linestyle="--",
    label="Theoretical probability = 0.5"
)

plt.xlabel("Number of trials")
plt.ylabel("Estimated probability of heads")
plt.title("Monte Carlo Convergence: Fair Coin Tosses")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Interpretation of the Coin-Toss Experiment

The graph illustrates how the estimated probability of obtaining heads changes as the number of simulated coin tosses increases.

For a small number of trials, the estimated probability can fluctuate considerably. As more tosses are performed, the influence of individual outcomes decreases, and the estimate generally becomes more stable around the theoretical probability of \(0.5\).

The experiment demonstrates that Monte Carlo simulation can approximate theoretical probabilities through repeated random sampling. However, the estimate does not have to move closer to the theoretical value after every additional trial.

This simple experiment provides a foundation for the more complex probability problems explored in the following sections.

### 2.7 The Law of Large Numbers

The **Law of Large Numbers** is a fundamental concept in probability theory. It states that, under appropriate assumptions, the sample average converges to the expected value as the number of independent observations increases.

In the coin-toss experiment, each toss can be represented by a random variable:

$$
X_i =
\begin{cases}
1, & \text{if the result is heads},\\
0, & \text{if the result is tails}.
\end{cases}
$$

For a fair coin, the probability of heads and tails is equal:

$$
P(X_i = 1) = P(X_i = 0) = \frac{1}{2}.
$$

The expected value of each toss is:

$$
\begin{aligned}
E[X_i]
&= 1 \cdot P(X_i = 1) + 0 \cdot P(X_i = 0)\\
&= 1 \cdot \frac{1}{2} + 0 \cdot \frac{1}{2}\\
&= \frac{1}{2} = 0.5.
\end{aligned}
$$

After \(N\) independent tosses, the estimated probability of obtaining heads is:

$$
\hat{p}_N = \frac{1}{N}\sum_{i=1}^{N} X_i.
$$

The Law of Large Numbers states that:

$$
\boxed{
\hat{p}_N \xrightarrow[N \to \infty]{\text{a.s.}} 0.5
}
$$

Here, *a.s.* means *almost surely*: the estimated probability converges to \(0.5\) with probability \(1\), assuming independent tosses of a fair coin.

This does not mean that the estimate must get closer to \(0.5\) after every toss, or that it must equal \(0.5\) after a finite number of tosses. Instead, it means that the estimate converges to the theoretical probability as the number of independent tosses increases.

### 2.8 Effect of Sample Size on Estimation Accuracy

The number of trials can influence the accuracy of a Monte Carlo probability estimate. To investigate this relationship, we examine the estimated probability of heads after different numbers of coin tosses.

For each sample size \(N\), we calculate the absolute error:

$$
\text{Absolute Error}(N) = |\hat{p}_N - p|
$$

where \(\hat{p}_N\) is the estimated probability and \(p = 0.5\) is the theoretical probability for a fair coin.

By comparing the errors at different sample sizes, we can investigate how estimation accuracy changes as more trials are performed.

The error is not guaranteed to decrease after every additional trial. However, larger sample sizes generally produce more stable estimates.

```python

# Choose sample sizes to compare
sample_sizes = np.array([10, 100, 500, 1_000, 5_000, 10_000])

# Count heads up to each selected sample size
cumulative_heads = np.cumsum(tosses)
heads_at_sizes = cumulative_heads[sample_sizes - 1]

# Calculate the estimated probability for each sample size
estimated_probabilities = heads_at_sizes / sample_sizes

# Calculate the absolute error
absolute_errors = np.abs(
    estimated_probabilities - theoretical_probability
)

# Print the results
print(f"{'Trials':>10} {'Estimate':>12} {'Absolute error':>16}")

for n, estimate, error in zip(
    sample_sizes, estimated_probabilities, absolute_errors
):
    print(f"{n:>10,} {estimate:>12.4f} {error:>16.4f}")
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    sample_sizes,
    absolute_errors,
    marker="o",
    label="Absolute error"
)

plt.xscale("log")
plt.xlabel("Number of trials (log scale)")
plt.ylabel("Absolute error")
plt.title("Effect of Sample Size on Monte Carlo Error")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

### Interpretation of the Sample Size Experiment

The experiment compares the estimated probability of heads at different sample sizes.

The absolute error measures the difference between the estimated probability and the theoretical probability of \(0.5\). A smaller absolute error indicates a more accurate estimate for the selected sample size.

The results illustrate that increasing the number of trials generally improves the stability of Monte Carlo estimates. However, the absolute error does not necessarily decrease at every sample size because random variation remains present.

This experiment provides practical evidence of the Law of Large Numbers and demonstrates why the number of simulated trials is an important consideration in probability estimation.

## 3. The Monty Hall Problem

### 3.1 Problem Description

The Monty Hall problem is a probability puzzle based on a game show with three closed doors. Behind one door is a car, while behind the other two doors are goats.

The contestant initially selects one door. The host, who knows what is behind every door, then opens one of the two unselected doors, always revealing a goat. The contestant is offered the opportunity to either keep the original choice or switch to the other unopened door.

The objective is to determine which strategy gives the contestant a higher probability of winning the car.

### Assumptions

The analysis is based on the following assumptions:

1. The car is equally likely to be behind any of the three doors.
2. The contestant selects one door at random.
3. The host knows the location of the car.
4. The host always opens an unselected door containing a goat.
5. The host always offers the contestant the opportunity to switch.

We compare two strategies:

- **Stay:** Keep the initially selected door.
- **Switch:** Choose the other unopened door after the host reveals a goat.

### 3.2 Mathematical Analysis

To analyze the Monty Hall problem, assume that the contestant initially selects Door 1. Since the car is equally likely to be behind any of the three doors, there are three possible locations for the car, each with probability \(1/3\).

| Car Location | Stay Strategy | Switch Strategy |
|---|---|---|
| Door 1 | Win | Lose |
| Door 2 | Lose | Win |
| Door 3 | Lose | Win |

The host always opens an unselected door containing a goat. Therefore, if the car is behind Door 2 or Door 3, switching leads to the car. If the car is behind Door 1, staying wins.

**Probability of winning by staying**

The contestant wins by staying only when the initial choice is correct. Since there are three equally likely car locations and only one matches the initial choice, the probability of winning by staying is

$$
P(\text{Stay wins}) = \frac{1}{3} \approx 0.3333.
$$

**Probability of winning by switching**

The contestant wins by switching whenever the initial choice is incorrect. This happens in two of the three possible car locations. Therefore,

$$
P(\text{Switch wins}) = \frac{2}{3} \approx 0.6667.
$$

**Conclusion**

Under the stated assumptions, switching gives a higher probability of winning the car than staying. The probability of winning by switching is twice the probability of winning by staying.


### 3.3 Monte Carlo Simulation

To investigate the Monty Hall problem experimentally, we use Monte Carlo simulation. Instead of calculating the outcome of a single game, we simulate a large number of independent games and estimate the probability of winning under two strategies: staying with the initial choice and switching after the host opens a door containing a goat.

For each strategy, the estimated probability of winning is calculated as the number of wins divided by the total number of simulated games.

We compare the simulated probabilities with the theoretical probabilities derived in Section 3.2.

```python

from src.monty_hall import simulate_monty_hall
import numpy as np

rng = np.random.default_rng()
n_simulations = 10000

stay_wins = sum(
    simulate_monty_hall(switch=False, rng=rng)
    for _ in range(n_simulations)
)

switch_wins = sum(
    simulate_monty_hall(switch=True, rng=rng)
    for _ in range(n_simulations)
)

stay_probability = stay_wins / n_simulations
switch_probability = switch_wins / n_simulations

print(f"Number of simulations: {n_simulations:,}")
print(f"Stay wins: {stay_wins:,}")
print(f"Stay estimated probability: {stay_probability:.4f}")
print(f"Switch wins: {switch_wins:,}")
print(f"Switch estimated probability: {switch_probability:.4f}")
```

### 3.4 Results and Interpretation

The Monte Carlo experiment simulated 10,000 games for each strategy.

The stay strategy resulted in 3,451 wins, giving an estimated winning probability of 0.3451. The switch strategy resulted in 6,656 wins, giving an estimated winning probability of 0.6656.

The theoretical probabilities are \(1/3 \approx 0.3333\) for staying and \(2/3 \approx 0.6667\) for switching.

The absolute errors were approximately 0.0118 for staying and 0.0011 for switching. Both estimates are reasonably close to their theoretical values, although the observed errors differ because the outcomes are random.

The experiment supports the theoretical conclusion that switching is the better strategy. In this simulation, switching produced substantially more wins than staying.

These results illustrate how Monte Carlo simulation can be used to investigate a probability problem and compare experimental estimates with theoretical predictions.

```python

import matplotlib.pyplot as plt
import numpy as np

strategies = ["Stay", "Switch"]

theoretical_probabilities = [1 / 3, 2 / 3]
simulated_probabilities = [stay_probability, switch_probability]

x = np.arange(len(strategies))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(
    x - width / 2,
    theoretical_probabilities,
    width,
    label="Theoretical probability"
)

plt.bar(
    x + width / 2,
    simulated_probabilities,
    width,
    label="Simulated probability"
)

plt.xticks(x, strategies)
plt.ylabel("Probability of winning")
plt.title("Monty Hall: Theoretical vs Simulated Probabilities")
plt.ylim(0, 1)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. The Birthday Paradox

### 4.1 Problem Description

The birthday paradox is a probability problem that asks how large a group of people must be for the probability that at least two people share the same birthday to exceed 50%.

At first, this probability may seem surprisingly high. Many people expect that a much larger group is required because there are 365 possible birthdays in a non-leap year.

However, the number of possible pairs of people increases rapidly as the group grows. Each pair represents another opportunity for two people to share a birthday.

### Assumptions

For this analysis, we use the following assumptions:

1. A year has 365 possible birthdays.
2. Leap years are ignored.
3. Each birthday is equally likely.
4. The birthdays of different people are independent.
5. We are interested in the probability that **at least two people** in the group share a birthday.

Our objective is to calculate this probability mathematically and then use Monte Carlo simulation to compare the theoretical result with an experimental estimate.

### 4.2 Mathematical Analysis

Let \(n\) be the number of people in a group.

The first person can have any of the 365 possible birthdays. For the second person to have a different birthday, there are 364 remaining possibilities. For the third person, there are 363 remaining possibilities, and so on.

Therefore, for \(n \leq 365\), the probability that all birthdays are different is

$$
P(\text{all different}) =
\prod_{k=0}^{n-1}\frac{365-k}{365}.
$$

Using the complement rule, the probability that at least two people share a birthday is

$$
P(\text{at least one match}) =
1-\prod_{k=0}^{n-1}\frac{365-k}{365}.
$$

The birthday paradox is commonly associated with a group of 23 people, for which the probability of at least one shared birthday is greater than 50%.

In the next steps, we will calculate this probability using Python and investigate how it changes as the group size increases.

```python

from src.birthday_paradox import birthday_match_probability

for group_size in [2, 10, 20, 23, 30, 50]:
    probability = birthday_match_probability(group_size)
    print(
        f"Group size: {group_size:2d}, "
        f"Probability of a shared birthday: {probability:.4f}"
    )
```

```python

from src.birthday_paradox import (
    birthday_match_probability,
    simulate_birthday_match,
)
import numpy as np

rng = np.random.default_rng()
n_simulations = 10_000
group_size = 23

matches = sum(
    simulate_birthday_match(group_size, rng=rng)
    for _ in range(n_simulations)
)

simulated_probability = matches / n_simulations
theoretical_probability = birthday_match_probability(group_size)

print(f"Number of simulations: {n_simulations:,}")
print(f"Group size: {group_size}")
print(f"Groups with a shared birthday: {matches:,}")
print(f"Simulated probability: {simulated_probability:.4f}")
print(f"Theoretical probability: {theoretical_probability:.4f}")
print(
    f"Absolute error: "
    f"{abs(simulated_probability - theoretical_probability):.4f}"
)
```

```python
group_sizes = [5, 10, 15, 20, 23, 25, 30, 40, 50]

print(
    f"{'Group size':>12} "
    f"{'Theoretical':>14} "
    f"{'Simulated':>12}"
)

for size in group_sizes:
    theoretical = birthday_match_probability(size)

    matches = sum(
        simulate_birthday_match(size, rng=rng)
        for _ in range(n_simulations)
    )

    simulated = matches / n_simulations

    print(f"{size:>12} {theoretical:>14.4f} {simulated:>12.4f}")
```

```python

import matplotlib.pyplot as plt

theoretical_probabilities = [
    birthday_match_probability(size)
    for size in group_sizes
]

simulated_probabilities = []

for size in group_sizes:
    matches = sum(
        simulate_birthday_match(size, rng=rng)
        for _ in range(n_simulations)
    )
    simulated_probabilities.append(matches / n_simulations)

plt.figure(figsize=(10, 5))

plt.plot(
    group_sizes,
    theoretical_probabilities,
    marker="o",
    label="Theoretical probability"
)

plt.plot(
    group_sizes,
    simulated_probabilities,
    marker="s",
    linestyle="--",
    label="Monte Carlo simulation"
)

plt.axhline(
    y=0.5,
    linestyle=":",
    label="50% probability"
)

plt.axvline(
    x=23,
    linestyle=":",
    label="23 people"
)

plt.xlabel("Group size")
plt.ylabel("Probability of a shared birthday")
plt.title("Birthday Paradox: Theoretical vs Simulated Probability")
plt.ylim(0, 1)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Bayes' Theorem

### 5.1 Problem Description

Bayes' theorem describes how to update the probability of an event when new evidence becomes available.

In this case study, we consider a hypothetical medical test. The test can produce a positive or negative result, but it is not perfectly accurate. Some people who have the condition may receive a negative result, while some people without the condition may receive a positive result.

The key question is:

**If a person receives a positive test result, what is the probability that the person actually has the condition?**

The answer depends not only on the accuracy of the test but also on how common the condition is in the population.

### Assumptions

For this analysis, we define the following parameters:

- **Prevalence:** the proportion of the population that has the condition.
- **Sensitivity:** the probability of a positive test result given that the person has the condition.
- **False-positive rate:** the probability of a positive test result given that the person does not have the condition.

We will calculate the probability of having the condition after a positive test result using Bayes' theorem. We will then compare the theoretical result with a Monte Carlo simulation.

### 5.2 Mathematical Analysis

Bayes' theorem allows us to calculate the probability of a condition given a positive test result.

Let:

- \(A\) denote the event that a person has the condition.
- \(B\) denote the event that the test result is positive.

Bayes' theorem is

$$
P(A \mid B) =
\frac{P(B \mid A)P(A)}{P(B)}.
$$

The probability of a positive result can be expressed as

$$
P(B) =
P(B \mid A)P(A)
+
P(B \mid \neg A)P(\neg A).
$$

We use the following hypothetical parameters:

- Prevalence: \(P(A)=0.01\).
- Sensitivity: \(P(B \mid A)=0.99\).
- False-positive rate: \(P(B \mid \neg A)=0.05\).

Therefore,

$$
P(B) = 0.99 \cdot 0.01 + 0.05 \cdot 0.99
= 0.0594.
$$

The probability of having the condition given a positive test result is

$$
P(A \mid B) =
\frac{0.99 \cdot 0.01}{0.0594}
\approx 0.1667.
$$

Thus, under these assumptions, the probability of having the condition after a positive test result is approximately **16.67%**.

This example demonstrates that the probability of a condition after a positive test depends on both test accuracy and the prevalence of the condition in the population.
